Data cleaning

```python
students = [
    "  JOHN",
    "mary ",
    " PETER ",
    "GRACE",
    "  james"
]
```

Use a for loop and appropriate string methods to clean all the names and store the results in a new list.

```python
cleaned_students = []
for s in students:
    cleaned_students.append(s.strip().title())

print(cleaned_students)
```
